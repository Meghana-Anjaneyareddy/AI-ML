```python
import pandas as pd

titanic_train_data = pd.read_csv('../data/processed_train.csv')
titanic_train_data.shape, titanic_train_data.head()

titanic_test_data = pd.read_csv('../data/processed_test.csv')
titanic_test_data.shape, titanic_test_data.head()
```

```python
finalized_features = ['Pclass', 'Age', 'Fare', 'Family_On_Board']

finalized_training_table = titanic_train_data[finalized_features]
y_train = titanic_train_data['Survived']

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(finalized_training_table, y_train, test_size=0.1)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

for max_iter in [0, 100, 200, 500, 1000]:
    model = LogisticRegression(max_iter=max_iter)
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    print("Max Iterations:", max_iter)
    print(classification_report(y_test, y_pred))
    print(confusion_matrix(y_test, y_pred))
    print("M", max_iter, "-" * 50)
```
