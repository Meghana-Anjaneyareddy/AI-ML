```python
import pandas as pd

credit_risk_dataset = pd.read_csv('../data/Credit Risk/labeled_total_data_numeric.csv')
credit_risk_dataset.shape, credit_risk_dataset.head()
```

```python
from sklearn.model_selection import train_test_split

X = credit_risk_dataset[[col for col in credit_risk_dataset.columns if col not in ['CREDIT_RISK', 'ID']]]
y = credit_risk_dataset['CREDIT_RISK']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```
